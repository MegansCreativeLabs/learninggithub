Såhär kan du läsa koden rad för rad med kommmandot variable.readline

```python
with open("files/example.txt") as file:
    first_line = file.readline()
    second_line = file.readline()

    print(first_line)
    print(second_line)

# Den läser av filraden och fortsätter på nästa automatiskt. Vill jag gå tillbaka kan jag lägga in en rad med:

    file.seek(0)
    first_line_again = file.readline()
print(first_line_again)
```

Skillnaden mellan readline och readlines 

* readline() läser en rad i taget.
* Varje gång du anropar den fortsätter den till nästa rad.
* print() skriver ut varje variabel separat.

* readlines() läser alla rader samtidigt och lägger dem i en lista.
* for-loop om ag vill skriva ut allt rad för rad. 



```python
with open("files/example.txt") as file:
    lines = file.readlines()
    print(lines)

     for line in lines:
         print(line)
```

Lättare sätt skriva det på direkt om jag vill öppna alla filer och skriva ut dem är

```python
with open("files/example.txt") as file:
    for line in file:
         print(line)
       # print(line.rstrip())           Ger snyggare output utan mellanrummet \n
       # print(line, end = "")          Fungerar på samma sätt 
```
